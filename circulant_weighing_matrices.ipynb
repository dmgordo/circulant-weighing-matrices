# Circulant Weighing Matrices

A *circulant weighing matrix* $CW(n,k)$ is an $n\times n$ matrix with entries in $\{0,\pm1\}$ whose rows are cyclic shifts and which satisfies $WW^T=kI_n$. The weight $k$ is a square, written $k=s^2$. We will identify CW's with sets $P$ and $N$, where $P$ is the set of positions of $1$'s in the first row of the matrix, and $N$ is the position of $-1$'s.

A CW is *proper* if every shift of $P$ and $N$ has some element relatively prime to $n$, so that it is not just a blown-up version of a $CW(d,k)$ for some $d|n$.  We will only be concerned with the existence of proper CW's.

**Database convention:** keys use $n$ and $s$, so, for example, `CW(28,4)` mean $CW(28,16)$, not $CW(28,4)$. 

The database records existence and nonexistence information for orders up to 1000 and square weights below $20^2$. A status `Yes`, `No`, and `Open` indicates known existence, known nonexistence, and an unresolved case, respectively. A status of `All` means that all inequivalent CW's with those parameters are known and listed in the database.

This notebook provides access to the `cwm.json` database using **standard Python 3**. Neither SageMath nor NumPy is required; pandas is optional and is used only to display formatted tables.

The next cell loads the database.

```python
from cwm_code import *
```

The following code is an example of how to search for particular values or ranges of the parameters.  It searches for parameters $n$ and $k$ where $n \in [75,99]$ and a $CW(n,k)$ has not been ruled out, and shows the results in a formatted table.

```python
T = init_tab()
for M in cwm:
    n = get_n(M)
    k = get_k(M)
    if get_status(M) in ['All','Yes','Open']: # skip 'No' cases
        if n in range(75,100):
            if k in range(9,100):
                add_tab_entry(T,M)
show_tab(T)
```

## Viewing existence results and circulant weighing matrices

`get_cwm_data(n,s)` reports the status of $CW(n,s^2)$ and any recorded support pairs $P,N$. The second argument is the square root of the weight, **not** the weight itself. 

```python
get_cwm_data(13,3)
```

`cwm_table()` prints a table of existence results, following the style of Tan's *Group invariant weighing matrices* (Designs, Codes and Cryptography, 2018). Column $s$ corresponds to weight $s^2$. Symbols: `Y` = existence (`Yes` or `All`), `.` = proven nonexistence (`No`), `*` = open.  Remember like, unlike Strassler's or Tan's tables, we are only considering proper circulant weighing matrices.

## Handling circulant weighing matrices

`get_cwm(n,s,i)` returns the $i$th recorded representative as `[n,s,P,N]` (zero-based index). Here $P$ and $N$ give the positions of the $+1$ and $-1$ coefficients of the corresponding element of $\mathbb{Z}[C_n]$. This is a list representation and requires no SageMath.

```python
get_cwm(28,4,3)
```

`is_cwm(M)` independently checks that the supports are valid and that the periodic autocorrelations vanish at every nonzero shift, so the associated circulant matrix satisfies $WW^T=s^2 I_n$. It uses only standard Python.

```python
is_cwm(get_cwm(28,4,3))
```
